```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('netflix_titles.csv')
```

```python
df
```

```python
df.shape  #Initial Data Shape
```

```python
print("Missing Values:")
print(df.isnull().sum())
```

```python
df = df.drop_duplicates()
```

```python
df
```

```python
df['date_added'] = pd.to_datetime(df['date_added'], errors='coerce')
```

```python
print("\nData Types:\n", df.dtypes)
```

```python
sns.boxplot(data=df, x='release_year')
plt.title("Outlier Detection: Release Year")
plt.show()
```

```python
print("\nMissing Values Before Handling:\n", df.isnull().sum())
```

```python
# For categorical columns filling NaN Vales with Mode
for col in df.select_dtypes(include='object').columns:
    mode_val = df[col].mode()[0]
    df[col].fillna(mode_val, inplace=True)
```

```python
df
```

```python
if 'date_added' in df.columns:
    mode_date = df['date_added'].mode()[0]
    df['date_added'].fillna(mode_date, inplace=True)
```

```python
df
```

```python
print("\nMissing Values After Handling:\n", df.isnull().sum())
```

```python
df.to_csv('cleaned_netflix_titles.csv', index=False)
print("✅ Cleaned data saved to 'cleaned_netflix_titles.csv'")
```
